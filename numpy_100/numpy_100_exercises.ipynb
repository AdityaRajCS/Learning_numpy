#### 17. What is the result of the following expression?
```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

````python
print(0 * np.nan) # O/P - nan
                  # np.nan means "Not a Number". NaN represents an undefined
                  # or unavailable numeric value.
print(np.nan == np.nan) # O/P - False.
                        # NaN is not equal to anything, including itself.
print(np.inf > np.nan)  # O/P - False.
                        # np.inf represents positive infinity.
                        # Comparisons involving NaN return False.
print(np.nan - np.nan)  # O/P - nan
                        # Arithmetic operations involving NaN generally result in NaN.
print(np.nan in set([np.nan]))  # O/P - True
                                # Set membership can identify the same NaN object
                                # even though np.nan == np.nan is False.
print(0.3 == 3 * 0.1)  # O/P - False.
                       # Due to floating-point precision, 3 * 0.1 is not represented
                       # exactly as 0.3.
                       # np.isclose(0.3, 3 * 0.1) can be used to check approximate equality.
````

#### 22. Normalize a 5x5 random matrix.

````python
# Min-max normalization -
arr = np.random.randint(1,50,(5,5))
print(arr)
print()
arr = (arr - arr.min())/(arr.max() - arr.min())
print(arr)
print()

# Standardization (Z-score normalization) -
arr = np.random.randint(1,100,(5,5))
arr = (arr - np.mean(arr)) / np.std(arr)
print(arr)
````

#### 24. Multiply a 5x3 matrix by a 3x2 matrix (real matrix product).

#### 26. What is the output of the following script?
```python
# Author: Jake VanderPlas

print(sum(range(5),-1))
from numpy import *
print(sum(range(5),-1))
```


````python
print(sum(range(5), -1))    # O/P - 9
                            # Python's sum() → sum(iterable, start=0).
                            # The second argument (-1) is the starting value:
                            # 0 + 1 + 2 + 3 + 4 - 1 = 9.
from numpy import *         # Imports all public names from NumPy into the current namespace.
                            # This causes NumPy's sum() to shadow Python's built-in sum().
print(sum(range(5), -1))    # O/P - 10
                            # NumPy's sum() → equivalent to np.sum() → np.sum(a, axis=None, ...).
                            # Here, -1 is interpreted as axis=-1(last axis), not as a starting value.
print(builtin.sum(range(5), -1))    # O/P - 9 -> Uses Python's sum().
````
