#### 15. Create a 2d array with 1 on the border and 0 inside.

````python
arr = np.ones((5,6),dtype= int)
arr[1:-1,1:-1] = 0
print(arr)
print()
# Another way to do this is - 
arr = np.zeros((5,6),dtype= int)
arr[0, :] = arr[-1, :] = 1
arr[:, 0] = arr[:, -1] = 1
print(arr)
````

#### 17. What is the result of the following expression?
```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

````python
print(0 * np.nan) # O/P - nan
                  # np.nan means "Not a Number". NaN represents an undefined
                  # or unavailable numeric value.
print(np.nan == np.nan) # O/P - False.
                        # NaN is not equal to anything, including itself.
print(np.inf > np.nan)  # O/P - False.
                        # np.inf represents positive infinity.
                        # Comparisons involving NaN return False.
print(np.nan - np.nan)  # O/P - nan
                        # Arithmetic operations involving NaN generally result in NaN.
print(np.nan in set([np.nan]))  # O/P - True
                                # Set membership can identify the same NaN object
                                # even though np.nan == np.nan is False.
print(0.3 == 3 * 0.1)  # O/P - False.
                       # Due to floating-point precision, 3 * 0.1 is not represented
                       # exactly as 0.3.
                       # np.isclose(0.3, 3 * 0.1) can be used to check approximate equality.
````
