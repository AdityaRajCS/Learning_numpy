#### 17. What is the result of the following expression?
```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

````python
print(0 * np.nan) # O/P - nan
                  # np.nan means "Not a Number". NaN represents an undefined
                  # or unavailable numeric value.
print(np.nan == np.nan) # O/P - False.
                        # NaN is not equal to anything, including itself.
print(np.inf > np.nan)  # O/P - False.
                        # np.inf represents positive infinity.
                        # Comparisons involving NaN return False.
print(np.nan - np.nan)  # O/P - nan
                        # Arithmetic operations involving NaN generally result in NaN.
print(np.nan in set([np.nan]))  # O/P - True
                                # Set membership can identify the same NaN object
                                # even though np.nan == np.nan is False.
print(0.3 == 3 * 0.1)  # O/P - False.
                       # Due to floating-point precision, 3 * 0.1 is not represented
                       # exactly as 0.3.
                       # np.isclose(0.3, 3 * 0.1) can be used to check approximate equality.
````

#### 22. Normalize a 5x5 random matrix.


#### 27. Consider an integer vector Z, which of these expressions are legal?
```python
Z**Z
2 << Z >> 2
Z <- Z
1j*Z
Z/1/1
Z<Z>Z
```

````python
Z = 3
print(Z**Z)          # Legal. Calculates 3 raised to the power of 3. O/P - 27.
print(2 << Z >> 2)   # Legal. First calculates 2 << Z, then applies >> 2 to the result. O/P - 4.
print(Z <- Z)        # Legal. Interpreted as Z < (-Z). O/P - False.
print(1j * Z)        # Legal. Multiplies the complex number 1j by Z. O/P - 3j.
print(Z / 1 / 1)     # Legal. Calculates (Z / 1) / 1. O/P - 3.0.
print(Z < Z > Z)     # Legal. This is a chained comparison: (Z < Z) and (Z > Z). O/P - False.
````


#### 28. What are the result of the following expressions?
```python
np.array(0) / np.array(0)
np.array(0) // np.array(0)
np.array([np.nan]).astype(int).astype(float)
```

````python
print(np.array(0) / np.array(0))  # Equivalent to 0 / 0. O/P - nan. And raises a RuntimeWarning.
print(np.array(0) // np.array(0))  # Equivalent to 0 // 0. O/P - 0.
                    # NumPy's floor division by zero produces 0 for integer arrays and raises a RuntimeWarning.
np.array([np.nan]).astype(int).astype(float)
                                # O/P - *some float value that does not represent nan.*
                                # Converting NaN directly to an integer is invalid; NumPy uses
                                # the minimum representable integer value, with a RuntimeWarning.
                                # Converting that integer back to float doesn't  restore NaN.
````

### `NaN` and Integer Conversion

```python
np.array([np.nan]).astype(int).astype(float)
```

> **Remember:**
>
> - `NaN` can be represented by floating-point dtypes, but **not by integer dtypes**.
> - Converting `NaN` to `int` produces a `RuntimeWarning` and typically converts it to the minimum representable integer value.
> - Converting that integer back to `float` **does not restore `NaN`**.
>
> ```text
> [NaN] → astype(int) → [-9223372036854775808] → astype(float) → [-9.22337204e+18]
> ```

#### 33. How to get the dates of yesterday, today and tomorrow?

````python
import datetime     # python's datetime module
today = datetime.date.today()
print("Yesterday was :", today - datetime.timedelta(days=1))
print("Today is :",today)
print("Tomorrow will be :", today + datetime.timedelta(days=1))

# Another way to write above code  -
from datetime import date, timedelta  # only imports the functions needed
today = date.today()
print("Yesterday was :", today - timedelta(days=1))
print("Today is :",today)
print("Tomorrow will be :", today + timedelta(days=1))

# Another solution (without using Python's datetime module) -
yesterday = np.datetime64('today') - np.timedelta64(1, 'D')
today     = np.datetime64('today')
tomorrow  = np.datetime64('today') + np.timedelta64(1, 'D')
````
