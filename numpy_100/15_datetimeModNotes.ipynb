# `datetime` Module

Python's built-in `datetime` module provides classes and functions for working with **dates, times, time differences, and date/time formatting**.

```python
import datetime
```

---

## `date`

`datetime.date` represents a **calendar date** without a time.

### `date.today()`

Returns the current local date.

```python
today = datetime.date.today()
print(today)
```

Output:

```text
2026-08-13
```

The result is a `date` object:

```python
print(type(today))
```

Output:

```text
<class 'datetime.date'>
```

---

### Creating a specific date

```python
d = datetime.date(2026, 8, 13)
print(d)
```

Output:

```text
2026-08-13
```

The arguments are:

```text
date(year, month, day)
```

---

## `datetime`

`datetime.datetime` represents **both a date and a time**.

### `datetime.now()`

Returns the current local date and time.

```python
now = datetime.datetime.now()
print(now)
```

Example output:

```text
2026-08-13 21:15:32.123456
```

It contains:

```text
year - month - day
hour : minute : second : microsecond
```

---

### Creating a specific date and time

```python
dt = datetime.datetime(2026, 8, 13, 15, 30, 45)
print(dt)
```

Output:

```text
2026-08-13 15:30:45
```

The arguments are:

```text
datetime(year, month, day, hour, minute, second, microsecond)
```

---

## `timedelta`

`datetime.timedelta` represents a **duration or difference between dates/times**.

### Adding days

```python
today = datetime.date(2026, 8, 13)

tomorrow = today + datetime.timedelta(days=1)
print(tomorrow)
```

Output:

```text
2026-08-14
```

### Subtracting days

```python
yesterday = today - datetime.timedelta(days=1)
print(yesterday)
```

Output:

```text
2026-08-12
```

### Other units

```python
datetime.timedelta(
    days=2,
    hours=3,
    minutes=30
)
```

A `timedelta` can represent:

- `days`
- `seconds`
- `microseconds`
- `milliseconds`
- `minutes`
- `hours`
- `weeks`

### Finding the difference between two dates

```python
date1 = datetime.date(2026, 8, 13)
date2 = datetime.date(2026, 8, 1)

difference = date1 - date2
print(difference)
```

Output:

```text
12 days, 0:00:00
```

You can get the number of days using:

```python
print(difference.days)
```

Output:

```text
12
```

# `strptime()`

Converts a **string into a `datetime` object**.

```text
string → datetime
```

### Example

```python
s = "13-08-2026"

dt = datetime.datetime.strptime(s, "%d-%m-%Y")
print(dt)
```

Output:

```text
2026-08-13 00:00:00
```

The format string tells Python how to interpret the string:

| Code | Meaning |
|---|---|
| `%Y` | 4-digit year |
| `%m` | 2-digit month |
| `%d` | 2-digit day |
| `%H` | Hour (24-hour format) |
| `%M` | Minute |
| `%S` | Second |
| `%f` | Microsecond |

For example:

```python
s = "13/08/2026 15:30:45"

dt = datetime.datetime.strptime(
    s,
    "%d/%m/%Y %H:%M:%S"
)

print(dt)
```

Output:

```text
2026-08-13 15:30:45
```

> **Remember:** `strptime()` is useful when you receive a date/time as **text** and want to convert it into a `datetime` object so you can perform date/time operations on it.

---

# `strftime()`

Converts a **`datetime` object into a formatted string**.

```text
datetime → string
```

It is essentially the reverse of `strptime()`.

### Example

```python
dt = datetime.datetime(2026, 8, 13, 15, 30, 45)

s = dt.strftime("%d-%m-%Y")
print(s)
```

Output:

```text
13-08-2026
```

Another example:

```python
print(dt.strftime("%d/%m/%Y %H:%M:%S"))
```

Output:

```text
13/08/2026 15:30:45
```

### `strptime()` vs `strftime()`

```text
strptime()  → string → datetime
strftime()  → datetime → string
```

---

# `timestamp()`

Converts a `datetime` object into a **Unix timestamp**.

A Unix timestamp represents the number of seconds since:

```text
1970-01-01 00:00:00 UTC
```

### Example

```python
dt = datetime.datetime(2026, 8, 13)

print(dt.timestamp())
```

Output will be a number similar to:

```text
1786579200.0
```

The exact value can depend on the timezone interpretation of the `datetime` object.

The result is a floating-point number, so it can also contain fractions of a second.

```python
print(type(dt.timestamp()))
```

Output:

```text
<class 'float'>
```

---

# `fromtimestamp()`

Converts a Unix timestamp into a `datetime` object.

```text
Unix timestamp → datetime
```

### Example

```python
timestamp = 1786579200

dt = datetime.datetime.fromtimestamp(timestamp)

print(dt)
```

Output:

```text
2026-08-13 00:00:00
```

The result represents the timestamp in the **local timezone**.

### `timestamp()` ↔ `fromtimestamp()`

These perform opposite conversions:

```text
datetime
   ↓
timestamp()
   ↓
Unix timestamp
   ↓
fromtimestamp()
   ↓
datetime
```

A `date` or `datetime` object provides individual components.

```python
dt = datetime.datetime(2026, 8, 13, 15, 30, 45)

print(dt.year)
print(dt.month)
print(dt.day)
print(dt.hour)
print(dt.minute)
print(dt.second)
```

Output:

```text
2026
8
13
15
30
45
```

## NumPy Date/Time Types

NumPy has **two main date/time types**:

| Type | Purpose |
|---|---|
| `np.datetime64` | Represents a **date or date/time** |
| `np.timedelta64` | Represents a **duration or time interval** |

### `np.datetime64`

Represents a specific date or date/time.

```python
np.datetime64('2026-08-13')           # Date
np.datetime64('2026-08-13T15:30:00')  # Date + time
```

### `np.timedelta64`

Represents a duration.

```python
np.timedelta64(1, 'D')   # 1 day
np.timedelta64(3, 'h')   # 3 hours
```

Common units include:

```text
Y  → years
M  → months
W  → weeks
D  → days
h  → hours
m  → minutes
s  → seconds
ms → milliseconds
us → microseconds
ns → nanoseconds
```

### Date arithmetic

`datetime64` and `timedelta64` can be used together:

```python
today = np.datetime64('today')

yesterday = today - np.timedelta64(1, 'D')
tomorrow  = today + np.timedelta64(1, 'D')
```

> **Remember:**
>
> - `np.datetime64` → **date / date-time**
> - `np.timedelta64` → **duration**
> - NumPy does not have separate `np.date` or `np.time` types like Python's `datetime` module.
> - These types are particularly useful for **date/time arrays and vectorized operations**.

## Finding the Day of the Week

### Using Python's `datetime`

`datetime.date` provides `strftime()` to get the weekday name.

```python
import datetime

date = datetime.date(2016, 7, 15)

print(date.strftime("%A"))
```

Output:

```text
Friday
```

Common weekday format codes:

```text
%A → Full weekday name     → Friday
%a → Abbreviated name      → Fri
```

---

### Using NumPy

NumPy's `datetime64` does not have a direct `weekday()` method. We can calculate the weekday using a known reference date.

```python
date = np.datetime64('2016-07-15')

days = (date - np.datetime64('1970-01-05')) // np.timedelta64(1, 'D')

print(days % 7)
```

Output:

```text
4
```

`1970-01-05` was a **Monday**, so the numbers correspond to:

```text
0 → Monday
1 → Tuesday
2 → Wednesday
3 → Thursday
4 → Friday
5 → Saturday
6 → Sunday
```

Therefore:

```text
2016-07-15 → 4 → Friday
```

> **Remember:**
>
> - Python `datetime` → use `strftime("%A")` for the weekday name.
> - NumPy `datetime64` → calculate the number of days from a known reference date and use `% 7`.
> - For practical weekday calculations, Python's `datetime` approach is much simpler.
