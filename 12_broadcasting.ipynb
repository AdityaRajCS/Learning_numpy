```python
import numpy as np
```

```python
arr1 = np.arange(1,21).reshape(4,5)
arr2 = np.ones(20,dtype=np.int8).reshape(4,5)
print(arr1)
print()
print(arr2)
```

```python
arr1 + arr2     # same shape. No need for broadcasting.
```

```python
arr1 + 10       # different shapes. Compatible for broadcasting.
```

```python
# arr3 = np.ones(20,dtype=np.int8).reshape(5,4)
# arr2 + arr3  # different shapes. Not compatible to broadcast.
#             # ValueError: operands could not be broadcast together with shapes (4,5) (5,4).
```

```python
arr4 = np.arange(1,6)
print(arr4)
print()
print(arr2)
arr2 + arr4         # arr4 is broadcast to (4,5).
```

```python
arr5 = np.arange(1,7).reshape(3,2)
arr6 = np.linspace(10,60,6,dtype=int).reshape(3,2)
print(arr5)
print()
print(arr6)
arr5 + arr6
```

```python
# arr6 = arr6.reshape(2,3)  
# arr5 + arr6           # shapes (3,2) and (2,3) can not be broadcast together.
```

```python
arr7 = np.array([[1,2]])  
print(arr7.shape)
arr5 + arr7             # (3,2) and (1,2) do broadcast together.
```

```python
arr9 = np.array([[1,2,3]])
arr10 = arr9.reshape(3,1)
print(arr9)
print()
print(arr10)
print()
print(arr9.shape)
print()
print(arr10.shape) 
arr9 * arr10        # both arrays are broadcast to (3,3).
```

```python
print(arr1)
arr8 = arr1[1]
print(arr8)
arr1 > arr8         # broadcasting with comparision.
```
