````python
import numpy as np
````

````python
arr = np.array([1,2,3,4,5])
arr
````

````python
print(arr)
print(arr.shape)        # shape of array (dimension of array as tuple).
print(arr.ndim)         # dimension(axes) of array (1-D, 2-D, 3-D, etc).
print(arr.size)         # total number of elements in array.
print(arr.itemsize)     # size (in bytes) of 1 array element.
print(arr.dtype)        # data-type of array.
print(arr.nbytes)       # space taken by array in bytes.
print(arr.astype(float)) # returns a copy of array with all the elements represented as floating-points.
````

---
## `shape`

Returns the dimensions of an array as a tuple.

> **Remember:**
> - `(5,)` → 1D array with 5 elements.
> - `(2,3)` → 2 rows, 3 columns. (2D array).
> - `()` → Scalar (0D array).

---
## `ndim`

Returns the **number of dimensions (axes)** of an array.

> **Remember:**
>
> - `1` → 1D array (Vector).
> - `2` → 2D array (Matrix).
> - `3` → 3D array (Collection of matrices).
> - `0` → Scalar (Single value).

---

## `size`

Returns the **total number of elements** in an array.

> **Remember:**
>
> - `(5,)` → `size = 5`
> - `(2,3)` → `size = 6` (`2 × 3`)
> - `(2,2,2)` → `size = 8` (`2 × 2 × 2`)
> - `()` → `size = 1` (Scalar)

---

## `itemsize`

Returns the **size (in bytes) of one array element**.

> **Remember:**
>
> - `int64` → `8` bytes
> - `float64` → `8` bytes
> - `int32` → `4` bytes
> - `bool` → `1` byte
> - Depends only on the array's **dtype**, not its shape or size.
---

````python
arr2 = np.array([[1,2,3],[4,5,6]])
print(arr2)
print()
print(arr2.shape)
print(arr2.ndim)
print(arr2.size)
print(arr2.itemsize) # prints 8 because dtype is int64.
print(arr2.dtype)  # prints the dtype
````

````python
print(arr2[0].shape)
print(arr2[0].ndim)
print(arr2[0].size)
print(arr2[0].itemsize)
````

````python
print(arr2[1].shape)
print(arr2[1].ndim)
print(arr2[1].size)
print(arr2[1].itemsize)
````

````python
print(arr2[0,1].shape) # single element in array has 0 dimension. The empty tuple () means 0-dimensional (scalar).
print(arr2[0,1].ndim)
print(arr2[0,1].size)
print(arr2[0,1].itemsize)
````

## `dtype`

Returns the **data type** of the array elements.

> **Remember:**
>
> - `int64` → 64-bit integer (8 bytes)
> - `float64` → 64-bit floating-point (8 bytes)
> - `bool` → Boolean (`True`/`False`) (1 byte)
> - `str` / `<U...>` → Unicode string
> - All elements in a NumPy array have the **same `dtype`**.

````python
print(arr2)
arr2.dtype
````

````python
print(arr3)
arr3.dtype
````

### An important function/method related to dtype() - 
## `astype()`

Array method/ function that converts an array to a **new data type**.

**Syntax**

```python
array.astype(dtype)
```

> **Remember:**
>
> - Returns a **new array** (original remains unchanged).
> - `float → int` **truncates**, not rounds.
> - Common types: `int`, `float`, `bool`, `str`.
> - Use `dtype=` while creating an array; use `astype()` to convert an existing array.

### Examples

```python
arr.astype(int)
arr.astype(float)
arr.astype(bool)
arr.astype(str)

# NumPy-specific data types

arr.astype(np.int32)
arr.astype(np.int64)
arr.astype(np.float32)
arr.astype(np.float64)
```

````python
print(arr)
print(arr.astype(float))
print(arr)
````

````python
print(arr2)
arr2.astype(bool)
````

## `nbytes`

Returns the **total memory occupied** by the array (in bytes).

> **Remember:**
>
> - `nbytes = size × itemsize`
> - Depends on both the **number of elements** and the **data type**.
> - Useful for checking an array's memory usage.


````python
arr = np.array([[1,2,3], [4,5,6]])
print(arr)
print()
print(arr.size,arr.itemsize)
arr.nbytes # returns (size*itemsize)
````

````python
arr5 = np.array([[True,False],[False,False]])
print(arr5)
print()
print(arr5.size,arr5.itemsize)
arr5.nbytes # returns (size*itemsize)
````

---
> **Additional method:** It looks like an attribute (has no parentheses), but is actually a shorthand for the `transpose()` function.

## `T`

Returns the **transpose** of the array (rows ↔ columns).

> **Remember:**
>
> - Works only for arrays with `ndim >= 2`.  
(Or more accurately: Transpose works on 1-D arrays too, but a 1-D array remains unchanged because it has only one axis).
> - Transpose swaps rows and columns.
> - `shape (m, n) → (n, m)`

---

````python
arr6 = np.array([[1,2,3],[4,5,6]])
print(arr6)
print()
print(arr6.T)
````

````python
arr7 = np.array([1,2,3])
print(arr7)
print()
print(arr7.T)  # Does nothing.
````
