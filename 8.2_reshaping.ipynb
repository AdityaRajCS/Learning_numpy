# Axis Manipulation Functions

These functions **rearrange the axes (dimensions)** of an array **without changing its data**.

Unlike `reshape()`, they **do not change the number of elements or the shape arbitrarily**. Instead, they **reorder existing axes**.

---

# `transpose()`

Swaps the axes of an array.

For a **2D array**, it converts **rows into columns** and **columns into rows**.

## Syntax

```python
array.transpose(*axes)
```

For a 2D array:

```python
array.transpose()
```

---

## Example (2D)

```python
arr = np.array([[1,2,3],
                [4,5,6]])

arr.transpose()
```

Output

```text
[[1 4]
 [2 5]
 [3 6]]
```

Original Shape

```python
(2,3)
```

Transposed Shape

```python
(3,2)
```

---

## Example (3D)

```python
arr.shape
```

```python
(2,3,4)
```

```python
arr.transpose(2,1,0)
```

New Shape

```python
(4,3,2)
```

The numbers

```python
(2,1,0)
```

tell NumPy the **new order of the axes**.

Original

```text
Axis 0 → 2
Axis 1 → 3
Axis 2 → 4
```

After

```text
Axis 2 becomes Axis 0

Axis 1 remains Axis 1

Axis 0 becomes Axis 2
```

---

## Remember

- Swaps axes.
- For 2D arrays:
  - Rows ↔ Columns.
- Returns a **view** whenever possible.
- Original array is unchanged.

---

# `.T`

A shortcut for `transpose()`.

Instead of

```python
arr.transpose()
```

simply write

```python
arr.T
```

Example

```python
arr = np.array([[1,2],
                [3,4]])

arr.T
```

Output

```text
[[1 3]
 [2 4]]
```

---

## Difference

```python
arr.T
```

is equivalent to

```python
arr.transpose()
```

For an n-dimensional array, .T reverses the order of all axes. For 2-D arrays, this is equivalent to swapping rows and columns.

---

## Remember

- Attribute (not a method).
- Shortcut for transpose.
- Easier to write.

---
---

# `swapaxes()`

Swaps **only two specified axes**.

Unlike transpose,

it does **not** rearrange every axis.

## Syntax

```python
array.swapaxes(axis1, axis2)
```

---

## 2-D Example

Suppose:

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

arr.shape
# (2, 3)
```

```python
arr.swapaxes(0, 1)
```

Output:

```text
[[1, 4],
 [2, 5],
 [3, 6]]
```

New Shape:

```text
(3, 2)
```

Axis `0` and Axis `1` are exchanged.

---

## Higher-D Example

Suppose:

```python
arr.shape
# (2, 3, 4)
```

```python
arr.swapaxes(0, 2)
```

New Shape:

```text
(4, 3, 2)
```

Only:

```text
Axis 0
```

and

```text
Axis 2
```

are exchanged.

Axis 1 stays exactly where it is.

---

Another Example:

```python
arr.swapaxes(1, 2)
```

```text
(2, 3, 4)
    ↓
(2, 4, 3)
```

---

## Remember

- Swaps **exactly two axes**.
- Other axes remain unchanged.
- Returns a **view** of the array.

---

# `moveaxis()`

Moves one axis to a new position.

Unlike `swapaxes()`,

it **does not exchange two axes**.

It simply **moves one axis** and shifts the remaining axes accordingly.

## Syntax

```python
np.moveaxis(array, source, destination)
```

---

## 2-D Example

Suppose:

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

arr.shape
# (2, 3)
```

Move:
Axis 0 to the last position.

```python
np.moveaxis(arr, 0, 1)
```

New Shape:

```text
(3, 2)
```

Output:

```text
[[1, 4],
 [2, 5],
 [3, 6]]
```

Here, Axis 0 is moved to position 1, while Axis 1 shifts to position 0.

---

## Higher-D Example

Original Shape:

```text
(2, 3, 4)
```

Move:

```text
Axis 2
```

to the front.

```python
np.moveaxis(arr, 2, 0)
```

New Shape:

```text
(4, 2, 3)
```

Notice:

> The other axes **shift automatically** to fill the position left by the moved axis.

---

Another Example:

```python
np.moveaxis(arr, 0, -1)
```

Original:

```text
(2, 3, 4)
```

↓

New:

```text
(3, 4, 2)
```

Axis 0 moves to the last position.

---

## Remember

- Moves **one specified axis** to a new position.
- The remaining axes **shift automatically**.
- Unlike `swapaxes()`, it does not simply exchange two axes.
- Returns a **view** of the array.

---

---

# Comparison

| Function | Purpose |
|----------|---------|
| `transpose()` | Rearranges all axes. |
| `.T` | Shortcut for `transpose()`. |
| `swapaxes()` | Swaps two specified axes. |
| `moveaxis()` | Moves one axis to another position. |

---

# Summary

| Function | Changes Shape | Rearranges Axes | Returns |
|----------|---------------|-----------------|----------|
| `transpose()` | ✔ | All axes | View (usually) |
| `.T` | ✔ | All axes | View (usually) |
| `swapaxes()` | ✔ | Two axes | View (usually) |
| `moveaxis()` | ✔ | One axis | View (usually) |
