# Matplotlib — Basic Graph Plotting

## What is Matplotlib?

**Matplotlib** is a Python library used to **create graphs and visualizations**.

It is commonly used together with NumPy:

```text
NumPy
  ↓
Generate / manipulate numerical data
  ↓
Matplotlib
  ↓
Visualize the data
```

---

## Importing Matplotlib

The commonly used plotting module is `pyplot`.

```python
import matplotlib.pyplot as plt
```

- `matplotlib` → the plotting/visualization library.
- `pyplot` → module containing commonly used plotting functions.
- `plt` → alias used to access `pyplot`.

Example:

```python
import matplotlib.pyplot as plt

plt.plot(x, y)
```

> **Remember:**  
> `plt` is just a shorter name (alias) for `matplotlib.pyplot`.

---

## `plt.plot()`

Used to **plot a line graph**.

```python
plt.plot(x, y)
```

- `x` → values for the x-axis.
- `y` → corresponding values for the y-axis.
- Matplotlib connects the points with lines.

Example:

```python
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
```

---

## Plotting Only `y`

You can also provide only the y-values:

```python
y = np.array([2, 4, 6, 8, 10])

plt.plot(y)
```

Matplotlib automatically uses:

```text
x = [0, 1, 2, 3, 4]
```

So this is approximately equivalent to:

```python
plt.plot([0, 1, 2, 3, 4], y)
```

---

## `plt.figure()`

Creates a **figure (the overall container/window in which plots are drawn)**.

```python
plt.figure()
```

You can control its size using `figsize`:

```python
plt.figure(figsize=(5, 3))
```

- First value → width.
- Second value → height.
- Values are approximately in inches.

> **Remember:**  
> `figure()` controls the size/container of the graph; `plot()` draws the actual line.

---

## `plt.xlabel()` and `plt.ylabel()`

Used to **label the axes**.

```python
plt.xlabel("Time")
plt.ylabel("Distance")
```

- `xlabel()` → x-axis label.
- `ylabel()` → y-axis label.

---

## `plt.title()`

Adds a **title** to the graph.

```python
plt.title("Distance vs Time")
```

---

## `plt.grid()`

Adds **grid lines** to the graph.

```python
plt.grid()
```

Can also be explicitly enabled or disabled:

```python
plt.grid(True)
plt.grid(False)
```

---

## `label=` and `plt.legend()`

When plotting multiple lines, `label=` **assigns a name** to each line.

```python
plt.plot(x, y1, label="sin(x)")
plt.plot(x, y2, label="cos(x)")
```

`label=` stores the label, but does **not display it on the graph by itself**.

```python
plt.legend()
```

displays the assigned labels as a **legend**.

> **Remember:**
> - `label=` → assigns a label to a plotted line.
> - `plt.legend()` → displays those labels.
> - A **legend** is a key that explains what different lines/symbols represent.

---

## `plt.show()`

Explicitly displays the graph.

```python
plt.show()
```

It is useful for clearly indicating that the plotting is finished.

---

## Why Does Jupyter Sometimes Show `Line2D`?

```python
plt.plot(x, y)
```

returns a **`Line2D` object** representing the line created by Matplotlib.

If it is the last expression in a Jupyter cell, Jupyter may display:

```text
[<matplotlib.lines.Line2D at 0x...>]
```

> **Remember:**  
> Jupyter displays the value returned by the **last expression** in a cell.

You can suppress this output with:

```python
plt.plot(x, y);
```

## Number of Points and Graph Smoothness

When plotting a function, NumPy first creates sample points.

```python
x = np.linspace(-50, 50, 50)
y = np.cos(x)

plt.plot(x, y)
```

Using too few points can make the graph look **pointy/angular** because Matplotlib connects the sampled points with straight lines.

Increasing the number of points gives a smoother graph:

```python
x = np.linspace(-50, 50, 1000)
```

> **Remember:**  
> The number of points controls **how densely the function is sampled**.

---

## Range vs Number of Points

These are two different things:

```python
np.linspace(-10, 10, 1000)
```

- `-10 → 10` → **range** of x-values.
- `1000` → **number of sample points**.

For periodic functions such as `sin(x)` and `cos(x)`, a very large range can contain many cycles.

For example:

```python
x = np.arange(-1000, 1000)
y = np.sin(x)
```

contains hundreds of sine cycles, so the graph can look extremely compressed.

> **Remember:**
>
> **Number of points** → sampling density / smoothness.  
> **x-range** → how much of the function is being displayed.

---

## Basic Complete Example

```python
x = np.linspace(0, 2*np.pi, 100)

y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(5, 3))

plt.plot(x, y1, label="sin(x)")
plt.plot(x, y2, label="cos(x)")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Sine and Cosine")
plt.grid()
plt.legend()

plt.show()
```

## Basic Matplotlib Workflow

```text
Import Matplotlib
      ↓
Create / obtain data
      ↓
Create figure (optional)
      ↓
Plot data
      ↓
Add labels / title / grid / legend
      ↓
Display using plt.show()
```

| `plt.ylabel()` | Label y-axis |
| `plt.title()` | Add graph title |
| `plt.grid()` | Add grid lines |
| `plt.legend()` | Display line labels |

## Matplotlib + NumPy

A common pattern is:

```python
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.show()
```

Here:

- **NumPy** → generates `x` and calculates `y`.
- **Matplotlib** → takes those arrays and visualizes them.

> **Key idea:**  
> NumPy is doing the **numerical work**; Matplotlib is doing the **visualization**.
