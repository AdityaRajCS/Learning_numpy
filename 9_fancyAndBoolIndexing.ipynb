## `and` vs `&` vs `&&`

| Operator | Used in | Meaning |
|----------|---------|---------|
| `and` | Python | Logical AND for single boolean values |
| `&` | NumPy | Element-wise AND between boolean arrays |
| `&&` | C/C++ | Logical AND operator |

### NumPy Boolean Conditions

```python
arr[(arr > 10) & (arr % 2 == 0)]
```

> **Remember:**
>
> - Use `&` for **element-wise AND** with NumPy arrays.
> - Put each condition inside **parentheses**.
> - `and` cannot be used with NumPy boolean arrays.
> - `&&` is not a Python operator; it is used for logical AND in C/C++.

-----------

````python
arr[(arr>20) | (arr%5 == 0)] 
````

````python
arr[(arr>=15) | (arr%2 != 0)] 
````

## Element-by-Element Condition Matching

NumPy conditions are evaluated **element by element**.

```python
arr >= 15
```

Each element is checked separately and produces a corresponding `True` or `False`.

Example:

```python
arr = np.array([10, 15, 8, 21])

arr >= 15
# [False, True, False, True]
```

> **Remember:**
> Each boolean value corresponds to the element at the **same position** in the original array.

```python
arr[arr >= 15]
# [15, 21]
```

The `True` positions are selected; `False` positions are excluded.

### With multiple conditions

```python
arr[(arr >= 15) | (arr % 2 != 0)]
```

`|` performs **element-by-element OR**.

```text
(arr >= 15)      → [F, T, F, T]
(arr % 2 != 0)   → [F, T, F, T]
                    ↓  ↓  ↓  ↓
OR                → [F, T, F, T]
```

Result:

```python
[15, 21]
```

At each position, the element is selected if **at least one condition is True**.



---

````python
arr5 = np.arange(1,13).reshape(3,4)
print(arr5)
print()
print(arr5[((arr5 % 2) == 0)])
print()
arr5[((arr5 % 2) == 0)] = 0 # Replaces every even element in the original array with 0.
print(arr5)
````

**NOTE -**

> You can use the boolean mask not only to select elements, but also to modify all matching elements in-place.

***Boolean indexing can be used for both selecting and modifying elements.***
