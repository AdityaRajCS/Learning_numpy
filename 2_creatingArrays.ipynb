### `np.empty()`

Creates an array **without initializing its values**.

```python
np.empty((2, 3))
```

Possible output:

```python
array([[6.23042070e-307, 4.67296746e-307, 1.69121096e-306],
       [1.95819673e-306, 1.42413555e-306, 1.78019082e-306]])
```

> **Note:**
> - The values shown are **garbage values** (whatever happened to be present in memory).
> - They are **unpredictable** and **should not be relied upon**.
> - `np.empty()` is faster than `np.zeros()` because NumPy **does not initialize** the array elements.
> - Always assign values to an `empty` array before using it in computations.

## Other Array Creation Functions

- **`np.logspace()`**
  - Creates numbers that are evenly spaced on a **logarithmic scale**.
  - Example:
    ```python
    np.logspace(1, 3, 3)
    # Output: array([  10.,  100., 1000.])
    ```

- **`np.geomspace()`**
  - Creates numbers that form a **geometric progression** (constant ratio).
  - Example:
    ```python
    np.geomspace(1, 1000, 4)
    # Output: array([   1.,   10.,  100., 1000.])
    ```

- **`np.mgrid()`**
  - Creates dense coordinate grids using slice notation.
  - Mainly used for numerical computations and plotting.

- **`np.meshgrid()`**
  - Converts 1D coordinate arrays into coordinate matrices.
  - Commonly used for plotting surfaces, contours, and evaluating functions of two variables.

> **Note:** `logspace()`, `geomspace()`, `mgrid()`, and `meshgrid()` are more advanced functions. They are widely used in scientific computing, data visualization, and machine learning, but are not essential when first learning NumPy.
