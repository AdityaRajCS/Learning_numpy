## `builtins`

`builtins` is a Python module that contains Python's **built-in functions, constants, and exceptions**.

For example:

```python
import builtins

builtins.sum([1, 2, 3])      # 6
builtins.max([1, 5, 3])      # 5
builtins.min([1, 5, 3])      # 1
builtins.abs(-10)            # 10
```

Normally, these functions can be used directly without importing `builtins`:

```python
sum([1, 2, 3])
max([1, 5, 3])
```

### Why is `builtins` useful?

It provides a way to **explicitly access the original Python built-in** when its name has been shadowed by another function.

For example:

```python
from numpy import *

sum(range(5), -1)          # NumPy's sum()

builtins.sum(range(5), -1) # Python's built-in sum() → 9
```

> **Remember:**
>
> - `builtins.sum` → Python's built-in `sum()`.
> - `builtins.max` → Python's built-in `max()`.
> - `builtins.min` → Python's built-in `min()`.
> - `builtins` can be useful when a built-in name has been **shadowed** by another variable or function.
> - `from numpy import *` can shadow Python built-ins such as `sum()`, so prefer:
>
> ```python
> import numpy as np
> ```
>
> and use NumPy functions explicitly with the `np.` prefix.


## `flags.writeable`

`flags.writeable` is a NumPy array attribute that indicates whether the array's data can be modified.

### Make an array read-only

```python
arr = np.array([1, 2, 3, 4])

arr.flags.writeable = False
```

Now the array cannot be modified:

```python
arr[0] = 10
```

```text
ValueError: assignment destination is read-only
```

### Make it writable again

```python
arr.flags.writeable = True
```

After this, the array can be modified normally.

### Check whether an array is writable

```python
print(arr.flags.writeable)
```

Output:

```text
True
```

or:

```text
False
```

> **Note:** `flags.writeable` controls whether the **array's data can be changed**. It does not make the NumPy array object completely immutable.

**Quick reference:**

| Expression                    | Meaning                         |
| ----------------------------- | ------------------------------- |
| `arr.flags.writeable`         | Check whether `arr` is writable |
| `arr.flags.writeable = False` | Make `arr` read-only            |
| `arr.flags.writeable = True`  | Make `arr` writable             |
