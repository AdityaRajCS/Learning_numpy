````python
x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.figure(figsize=(5,3))
plt.plot(x, y1, label="sin(x)") # label doesn't show (reason is explained in the next markdown).
plt.plot(x, y2, label="cos(x)") # label doesn't show.
plt.show()
````

````python
x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.figure(figsize=(5,3))
plt.plot(x, y1, label="sin(x)")
plt.plot(x, y2, label="cos(x)") 
plt.legend()                    # Displays the label.
plt.show()
````

---

## `label=` vs `plt.legend()`

```python
plt.plot(x, y1, label="sin(x)")
```

**does store the label**, but it does **not display the label on the graph by itself**.

The `label=` argument tells Matplotlib:

> "Give this plotted line the name `sin(x)`."

Then:

```python
plt.legend()
```

tells Matplotlib:

> "Actually display the labels of the plotted lines as a legend."

Without `plt.legend()`, the labels are still **attached to the plotted lines**, they're just **not displayed**.


>*It's called a `legend` because in graphs, a legend is a key that explains what the symbols, lines, or colors represent.*
