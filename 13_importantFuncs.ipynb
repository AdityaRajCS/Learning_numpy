## `np.random.seed()`

```python
np.random.seed(1)
```

Sets the **starting state of the random number generator**.

Using the same seed produces the **same sequence of random numbers** each time.

> **Remember:**  
> `seed()` does not fix one random number; it makes the **entire sequence reproducible**.

### Choosing a Seed

The seed can be any **non-negative integer**:

```python
np.random.seed(0)      # valid
np.random.seed(1)      # valid
np.random.seed(1000)   # valid
```

Different seeds generally produce different sequences:

```text
seed(0)    → Sequence A
seed(1)    → Sequence B
seed(42)   → Sequence C
```

But using the **same seed** again reproduces the same sequence.

```python
np.random.seed(1000)
```

Running the same code again with `seed(1000)` produces the same sequence of random numbers.

### Negative Seeds

Negative integers are **not valid** for `np.random.seed()`:

```python
np.random.seed(-1)     # ❌
```

So:

```text
0 or positive integer → ✅
negative integer      → ❌
```

````python
np.random.seed(10)      # seeding to 10.
arr2 = np.random.randint(0,100,20)
print(arr2)
print()
print(np.where(arr2%2 != 0,-1,arr2))    # Returns new array with all odd ints replaced by -1.
print()                                 # Does not modify the original array.
print(arr2)
````

## Why `np.where(condition)` Returns a Tuple

When `np.where()` is used with **only a condition**:

```python
np.where(condition)
```

it returns a **tuple containing arrays of indices**, rather than an array directly.

For a 1-D array:

```python
arr = np.array([5, 8, 3, 10])

np.where(arr > 5)
```

returns:

```python
(array([1, 3]),)
```

The comma is important:

```python
(array([1, 3]),)
```

means **a tuple containing one array**.

### Why a Tuple?

`np.where()` also works with multidimensional arrays.

For example:

```python
arr = np.array([[5, 8, 3],
                [10, 2, 7]])

np.where(arr > 5)
```

returns:

```python
(array([0, 1, 1]), array([1, 0, 2]))
```

- First array → row indices
- Second array → column indices

Together, they identify the positions where the condition is `True`:

```text
(0, 1) → 8
(1, 0) → 10
(1, 2) → 7
```

> **Remember:**  
> The number of index arrays returned by `np.where(condition)` corresponds to the **number of dimensions** of the input array.

For a 1-D array → one index array:

```python
(array([1, 3]),)
```

For a 2-D array → two index arrays:

```python
(array([0, 1, 1]), array([1, 0, 2]))
```

To extract the index array from a 1-D result:

```python
np.where(arr > 5)[0]
```

returns:

```python
array([1, 3])
```

## `np.where(condition, x, y)`

The 3-argument form of `np.where()` works like an **element-by-element if-else**.

```python
np.where(condition, x, y)
```

- `condition` → condition to check for each element.
- `x` → value selected when the condition is `True`.
- `y` → value selected when the condition is `False`.

Example:

```python
arr = np.array([5, 8, 3, 10])

np.where(arr > 5, -1, arr)
```

Output:

```text
[ 5 -1  3 -1]
```

Conceptually:

```text
arr > 5
   ↓
5  → False → use arr value →  5
8  → True  → use -1         → -1
3  → False → use arr value →  3
10 → True  → use -1         → -1
```

So:

```python
np.where(condition, x, y)
```

can be thought of as:

```text
if condition is True  → x
if condition is False → y
```

> **Remember:**  
> The operation is performed **element by element** and `np.where()` returns a **new array**; it does not modify the original array in-place.
