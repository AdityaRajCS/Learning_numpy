### Why does `arr /= 2` give an error?

```python
arr /= 2
```

- `arr /= 2` means: **"Divide each element by 2 and store the result back into the original array."**
- If `arr` has `dtype=int64`, it can only store integers.
- However, `/` always produces **floating-point (`float64`)** values.
- NumPy does not automatically cast the `float64` result back to `int64`, so it raises a `UFuncTypeError`.

**This works:**

```python
arr = arr / 2
```

- A **new array** is created.
- The new array has `dtype=float64`, so it can store the floating-point results.

> **Remember:**
>
> - `arr /= 2` → Modifies the original array (**fails for `int64` arrays**).
> - `arr = arr / 2` → Creates a **new `float64` array** (**works**).
> - `arr //= 2` → Performs integer (floor) division and works with `int64` arrays.

````python
arr = np.arange(10)
print(arr)
arr //= 2 # Sol 1 to above prob: Use floor division. Integer results can be stored back in the original int array.
arr
````

````python
arr = np.arange(10) 
print(arr)
arr = arr / 2  # Solution 2: Create a new float64 array to store the result.
arr
````

````python
arr = np.arange(10)
print(arr)
arr = arr.astype(float) # Solution 3: Convert the original array to float64
arr /= 2          # Now in-place division works because the array can store float values.
arr
````

**NOTE - 
`arr = arr.astype(float)`

**Creates** a new float64 array and assigns it to arr.

## Matrix Multiplication

NumPy provides several ways to perform matrix multiplication.

### 1. `@` Operator — Recommended

```python
A @ B
```

- Specifically designed for matrix multiplication.
- Most readable and recommended for matrix operations.
- Equivalent to `np.matmul(A, B)`.

### 2. `.dot()` Method

```python
A.dot(B)
```

- Performs NumPy's `dot` operation.
- For 2D arrays, it is equivalent to matrix multiplication.
- It is the **method form** of the `dot` operation.

### 3. `np.matmul()` Function

```python
np.matmul(A, B)
```

- Specifically performs matrix multiplication.
- Equivalent to `A @ B`.
- For 2D arrays, it gives the same result as `np.dot(A, B)`.

### 4. `np.dot()` Function

```python
np.dot(A, B)
```

- Performs NumPy's `dot` operation.
- For 2D arrays, it is equivalent to matrix multiplication.
- It is the **function form** of the `dot` operation.

### `np.dot()` vs `.dot()`

Both perform the same NumPy `dot` operation.

```python
np.dot(A, B)   # Function form
A.dot(B)       # Method form
```

The main difference is how they are accessed:

- `np.dot(A, B)` → `dot()` function from the NumPy module.
- `A.dot(B)` → `dot()` method of the NumPy array `A`.

For 2D arrays:

```python
np.dot(A, B) == A.dot(B)
```

### Example

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])
```

All four produce the same result for these 2D arrays:

```python
A @ B

A.dot(B)

np.matmul(A, B)

np.dot(A, B)
```

Output:

```python
array([[19, 22],
       [43, 50]])
```

### Remember

| Operator / Function | Operation |
|---|---|
| `A * B` | Element-wise multiplication |
| `A @ B` | Matrix multiplication |
| `np.matmul(A, B)` | Matrix multiplication |
| `A.dot(B)` | Dot product; matrix multiplication for 2D arrays |
| `np.dot(A, B)` | Dot product; matrix multiplication for 2D arrays |

> **Tip:** Prefer `@` when you specifically mean matrix multiplication. It makes your intention clear.

## `axis` in NumPy

For a 2-D array:

```text
shape = (rows, columns)
        axis 0   axis 1
```

> **Remember:**  
> `axis` tells NumPy **which dimension to collapse/operate along**, not whether the result is a row or column.

```text
axis=0 → collapse rows    → one result per column
axis=1 → collapse columns → one result per row
```

Example:

```python
arr.std(axis=0)  # std of each column
arr.std(axis=1)  # std of each row
```
